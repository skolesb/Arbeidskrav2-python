```python
# Oppgave 5 - Areal og omkrets av trekant og halvsirkel

import numpy as np

a = float(input("Skriv inn lengden a: "))
b = float(input("Skriv inn lengden b: "))
```

```python
# Oppgave 5 - Beregning av areal og ytre omkrets

import numpy as np

# Funksjon som beregner areal og omkrets
def beregn_figur(a, b):

    # Beregner radius til halvsirkelen
    radius = a / 2

    # Beregner arealet av trekanten
    areal_trekant = a * b / 2

    # Beregner arealet av halvsirkelen
    areal_halvsirkel = np.pi * radius**2 / 2

    # Beregner samlet areal
    areal = areal_trekant + areal_halvsirkel

    # Beregner den skrå siden med Pytagoras
    skra_side = np.sqrt(a**2 + b**2)
        # Beregner lengden av halvsirkelens bue
    sirkelbue = np.pi * radius

    # Beregner den ytre omkretsen
    omkrets = b + skra_side + sirkelbue

    # Returnerer resultatene
    return areal, omkrets
```

```python
# Tester funksjonen med a = 4 og b = 6
a = 4
b = 6

# Kaller funksjonen
areal, omkrets = beregn_figur(a, b)

# Skriver ut resultatene
print(f"Figurens areal er {areal:.2f}")
print(f"Figurens ytre omkrets er {omkrets:.2f}")
```

```python
# Oppgave 6 - Tegne grafen til f(x) = -x^2 - 5

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Lager 100 x-verdier mellom -10 og 10
x = np.linspace(-10, 10, 100)
```

```python
# Beregner y-verdiene til funksjonen
y = -x**2 - 5
```

```python
# Tegner grafen
plt.plot(x, y)

# Legger til tittel og navn på aksene
plt.title("Grafen til f(x) = -x² - 5")
plt.xlabel("x")
plt.ylabel("f(x)")

# Viser rutenett
plt.grid(True)

# Viser grafen
plt.show()
```
